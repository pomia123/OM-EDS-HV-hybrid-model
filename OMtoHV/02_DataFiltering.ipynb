```python
"""
Microstructure Feature Filtering Using Statistical Z-Score Analysis.

This script removes statistically abnormal samples from the extracted
microstructural feature dataset using a predefined Z-score threshold.
The filtered dataset is subsequently exported for machine learning analysis.
"""

import pandas as pd
import numpy as np
from scipy.stats import zscore

# =============================================================================
# 1. Load Dataset & Define Target Feature Columns
# =============================================================================

df = pd.read_csv(r'./data/b_hv_with_features.csv')

# Handle missing descriptors carefully before statistical profiling
df.fillna(0, inplace=True)

drop_cols = ['FILE_NAME', 'HV']
feature_cols = [c for c in df.columns if c not in drop_cols]

# =============================================================================
# 2. Outlier Detection via Standard Z-Score Analysis
# =============================================================================

zscore_threshold = 3.0

# Calculate absolute Z-scores across all microstructural features
z_scores = np.abs(zscore(df[feature_cols], nan_policy='omit'))

# Filter rows where all feature metrics fall within the threshold boundary
mask = (z_scores < zscore_threshold).all(axis=1)
df_clean = df[mask].copy().reset_index(drop=True)

# =============================================================================
# 3. Export Cleaned Dataset & Logging
# =============================================================================

output_path = r'./data/c_cleaned_hv_with_features.csv'
df_clean.to_csv(output_path, index=False)

# Log filtering statistics matching pipeline conventions
print(f"\n[INFO] Outlier filtering completed successfully.")
print(f"[INFO] Original samples: {len(df)} | Remaining samples: {len(df_clean)} | Removed samples: {len(df) - len(df_clean)}")
print(f"[INFO] Filtered dataset successfully saved to: {output_path}")
```
